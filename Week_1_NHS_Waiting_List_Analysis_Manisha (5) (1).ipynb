# Section D – Business Analysis

## Business Question

### Which specialty has the longest median waiting time, and has it worsened in the last six months compared to the previous six months?

This analysis aims to identify the medical specialty with the highest median waiting time and determine whether waiting times have increased during the most recent six-month period compared to the previous six months. The findings will help the NHS identify departments requiring additional resources, improved capacity planning, and operational improvements.

### Step 1: Convert the Referral Date into Date Format

The referral date is converted into a date format so that the dataset can be divided into two six-month periods for comparison.

```python
df["referral_date"] = pd.to_datetime(df["referral_date"])

df["referral_date"].head()
```

### Step 2: Identify the Latest Referral Date

The most recent referral date is identified so that the last six months and the previous six months can be calculated accurately.

```python
latest_date = df["referral_date"].max()

latest_date
```

### Step 3: Divide the Dataset into Two Six-Month Periods

The dataset is divided into two periods:
- Current six months
- Previous six months

This comparison helps determine whether waiting times have increased or decreased over time.

```python
current_period = df[
    df["referral_date"] >= latest_date - pd.DateOffset(months=6)
]

previous_period = df[
    (df["referral_date"] < latest_date - pd.DateOffset(months=6)) &
    (df["referral_date"] >= latest_date - pd.DateOffset(months=12))
]
```

### Step 4: Calculate the Median Waiting Time by Specialty

The median waiting time is calculated for each medical specialty in both six-month periods. The median is used because it is less affected by extreme values and provides a more reliable measure of the typical waiting time.

```python
current_median = (
    current_period.groupby("specialty")["wait_days"]
    .median()
)

previous_median = (
    previous_period.groupby("specialty")["wait_days"]
    .median()
)
```

### Step 5: Compare Median Waiting Times

The median waiting times from the current six-month period are compared with those from the previous six-month period to determine whether waiting times have improved or worsened.

```python
comparison = pd.DataFrame({
    "Previous_6_Months": previous_median,
    "Current_6_Months": current_median
})

# Replace missing values with 0
comparison = comparison.fillna(0)

# Calculate the change in median waiting time
comparison["Change"] = (
    comparison["Current_6_Months"] -
    comparison["Previous_6_Months"]
)

# Sort by the largest increase
comparison = comparison.sort_values(
    by="Change",
    ascending=False
)

comparison
```

### Step 6: Visualise the Change in Median Waiting Time

A bar chart is created to compare the change in median waiting time across different medical specialties. This helps identify which specialties have experienced the greatest increase in waiting times.

```python
comparison["Change"].plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Change in Median Waiting Time by Specialty")
plt.xlabel("Specialty")
plt.ylabel("Change in Median Waiting Days")

plt.xticks(rotation=45)

plt.show()
```

## Observation

The bar chart compares the change in median waiting time across different medical specialties. Some specialties show higher median waiting times than others, indicating differences in patient demand and service capacity.

## Insight

The analysis suggests that certain specialties are experiencing greater pressure, resulting in longer waiting times. These departments may require additional staffing, specialist support, or increased appointment capacity to improve patient flow and reduce delays.

## Business Recommendation

Based on the analysis, the NHS should prioritise specialties with the highest median waiting times by improving workforce planning, allocating additional clinical resources, and expanding service capacity. Regular monitoring of waiting time trends will support better operational decision-making and improve patient outcomes.
